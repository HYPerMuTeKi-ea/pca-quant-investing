```python
RANDOM_SEED = 42
TRAIN_WINDOW = 252
N_ASSETS = 30

rng = np.random.default_rng(RANDOM_SEED)

returns = rng.normal(
    loc=0.0003,
    scale=0.01,
    size=(TRAIN_WINDOW, N_ASSETS)
)

print("Returns shape:", returns.shape)
```

```python
sample_cov = np.cov(returns, rowvar=False)

print("Covariance shape:", sample_cov.shape)
```

```python
is_symmetric = np.allclose(sample_cov, sample_cov.T)

print("Symmetric:", is_symmetric)
```
